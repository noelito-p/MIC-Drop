# 01 — Model download

This notebook downloads the pretrained models used later for feature extraction.

Outputs are saved under:

```text
<PROJECT_ROOT>/Output/
    pretrained_model_cache/
    downloaded_pretrained_models/
        resnet50_imagenet.pth
        efficientnet_b0_imagenet.pth
        densenet121_imagenet.pth
        mobilenet_v3_large_imagenet.pth
        vit_b_16_imagenet.pth
        cytoimagenet/efficientnetb0_weights-notop.h5
```

````python
from pathlib import Path
import os
import torch
from torchvision import models
import urllib.request

# Change this path if your shared folder is different.
PROJECT_ROOT = Path(r"..")
OUTPUT_DIR = PROJECT_ROOT / "output"

MODEL_DIR = OUTPUT_DIR / "downloaded_pretrained_models"
TORCH_CACHE_DIR = OUTPUT_DIR / "pretrained_model_cache"

MODEL_DIR.mkdir(parents=True, exist_ok=True)
TORCH_CACHE_DIR.mkdir(parents=True, exist_ok=True)

os.environ["TORCH_HOME"] = str(TORCH_CACHE_DIR)
torch.hub.set_dir(str(TORCH_CACHE_DIR))

print("Output folder:", OUTPUT_DIR)
print("Model folder:", MODEL_DIR)
print("Torch cache folder:", torch.hub.get_dir())
````

````python
TORCH_MODEL_SPECS = {
    "resnet50": {
        "builder": models.resnet50,
        "weights": models.ResNet50_Weights.DEFAULT,
    },
    "efficientnet_b0": {
        "builder": models.efficientnet_b0,
        "weights": models.EfficientNet_B0_Weights.DEFAULT,
    },
    "densenet121": {
        "builder": models.densenet121,
        "weights": models.DenseNet121_Weights.DEFAULT,
    },
    "mobilenet_v3_large": {
        "builder": models.mobilenet_v3_large,
        "weights": models.MobileNet_V3_Large_Weights.DEFAULT,
    },
    "vit_b_16": {
        "builder": models.vit_b_16,
        "weights": models.ViT_B_16_Weights.DEFAULT,
    },
}

print("Models to download:")
for name in TORCH_MODEL_SPECS:
    print("-", name)
````

````python
downloaded_models = {}

for model_name, spec in TORCH_MODEL_SPECS.items():
    save_path = MODEL_DIR / f"{model_name}_imagenet.pth"

    print("=" * 80)
    print(f"Loading pretrained {model_name}...")

    model = spec["builder"](weights=spec["weights"])
    model.eval()

    downloaded_models[model_name] = model

    if save_path.exists():
        print("Already saved:", save_path)
    else:
        torch.save(model.state_dict(), save_path)
        print("Saved:", save_path)

print("=" * 80)
print("Finished downloading/loading ImageNet pretrained models.")
````

````python
# CytoImageNet EfficientNetB0 weights.
# This is a Keras/TensorFlow EfficientNetB0 backbone weight file without the classification head.

cyto_model_dir = MODEL_DIR / "cytoimagenet"
cyto_model_dir.mkdir(parents=True, exist_ok=True)

cyto_weights_path = cyto_model_dir / "efficientnetb0_weights-notop.h5"

cyto_url = (
    "https://raw.githubusercontent.com/stan-hua/CytoImageNet/"
    "master/model_weights/efficientnetb0_weights-notop.h5"
)

if cyto_weights_path.exists():
    print("Already exists:", cyto_weights_path)
else:
    print("Downloading CytoImageNet EfficientNetB0 weights...")
    urllib.request.urlretrieve(cyto_url, cyto_weights_path)
    print("Downloaded:", cyto_weights_path)

print("Finished.")
````
